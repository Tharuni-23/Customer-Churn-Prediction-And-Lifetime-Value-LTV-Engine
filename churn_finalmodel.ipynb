```python
from sqlalchemy import create_engine, text

engine = create_engine(
    "postgresql+psycopg2://postgres:2602@[::1]:5432/postgres"
)

try:
    with engine.connect() as conn:
        print("CONNECTED")

        result = conn.execute(
            text("""
                SELECT datname
                FROM pg_database
                ORDER BY datname;
            """)
        )

        print("Databases seen by Python:")
        for row in result:
            print(row[0])

except Exception as e:
    print("ERROR:")
    print(e)
```
