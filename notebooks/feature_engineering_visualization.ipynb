```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("../data/processed/Feature_Engineered_Telco_Customer_Churn.csv")

df.head()
```

```python
new_features = [
    "TenureGroup",
    "TotalServices",
    "AverageMonthlySpend",
    "ExpectedCharges",
    "ChargeDifference",
    "ServiceUsageRatio",
    "HighValueCustomer",
    "ShortTenureCustomer"
]

df[new_features].head()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="TenureGroup",
    hue="Churn"
)

plt.title("Customer Churn by Tenure Group")
plt.xlabel("Tenure Group")
plt.ylabel("Number of Customers")
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="TotalServices",
    hue="Churn"
)

plt.title("Customer Churn by Total Number of Services")
plt.xlabel("Total Services")
plt.ylabel("Number of Customers")
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="AverageMonthlySpend"
)

plt.title("Average Monthly Spend by Churn")
plt.xlabel("Churn")
plt.ylabel("Average Monthly Spend")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="ExpectedCharges"
)

plt.title("Expected Charges by Churn")
plt.xlabel("Churn")
plt.ylabel("Expected Charges")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="ChargeDifference"
)

plt.title("Charge Difference by Churn")
plt.xlabel("Churn")
plt.ylabel("Charge Difference")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Churn",
    y="ServiceUsageRatio"
)

plt.title("Service Usage Ratio by Churn")
plt.xlabel("Churn")
plt.ylabel("Service Usage Ratio")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="HighValueCustomer",
    hue="Churn"
)

plt.title("Customer Churn by High-Value Customer Status")
plt.xlabel("High-Value Customer")
plt.ylabel("Number of Customers")
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="ShortTenureCustomer",
    hue="Churn"
)

plt.title("Customer Churn by Short-Tenure Customer Status")
plt.xlabel("Short-Tenure Customer")
plt.ylabel("Number of Customers")
plt.legend(title="Churn")
plt.tight_layout()
plt.show()
```

## Feature Engineering Insights

### 1. Tenure Group
- Churn patterns vary across different tenure groups.
- Shorter-tenure customers generally show higher churn.

### 2. Total Services
- Churn varies with the number of services used by customers.
- Customers with fewer services show different churn patterns compared with customers using more services.

### 3. Average Monthly Spend
- Average monthly spending differs between churned and non-churned customers.
- Spending level can be considered when analyzing customer churn.

### 4. Expected Charges
- Expected charges show differences between churned and non-churned customers.
- This feature can help identify customer spending patterns.

### 5. Charge Difference
- Charge difference varies between churned and non-churned customers.
- This feature may provide additional information about customer billing behavior.

### 6. Service Usage Ratio
- Service usage ratio shows differences across churn groups.
- It provides an additional measure of how customers use available services.

### 7. High-Value Customer
- Churn patterns can be compared between high-value and other customers.
- This feature helps identify whether customer value is associated with churn.

### 8. Short-Tenure Customer
- Short-tenure customers can be compared with other customers based on churn.
- This feature provides an additional indicator of early customer attrition.

### Overall Observation
- The engineered features provide additional perspectives on customer tenure, service usage, spending, billing behavior, and customer value.
- These features can be used in the next stage of the project for classification models and churn prediction.
